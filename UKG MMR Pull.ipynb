```python
#perform imports - json, pandas, and pathlib modules
import json
import pandas as pd
from pathlib import Path
```

```python
#read json file
file = (r'C:\Users\alysha.wood\OneDrive - Six Flags Entertainment Corp\Python Data\TST MMR API.json')
pd.read_json(file, typ='series')
```

```python
# loading the file into JSON
with open(file, encoding='utf-8') as f:
    d = json.load(f)

print(d)
```

```python
#create list for employee IDs
ids = []
for item in d:
    ids.append(item['personIdentity']['personNumber'])
print(ids)
```

```python
#create list for roles
roles = []
for item in d:
    roles.append(item['roleAssignments'][-1]['name'])

print(roles)
```

```python
#IDs and roles together (print)
print(ids, roles)
```

```python
#creating a dataframe with IDs and roles lists created
df = pd.DataFrame(roles, ids)

df
```

```python
#create dataframe with roles that do not equal Initial Role into a variable
filteredData = df.loc[df[0] != 'Initial Role']
filteredData
```

```python
#export dataframe data to csv file
filepath = Path('UKG MMR Audit/Audit Data/UKG MMR Audit 8.13.26.csv')
filepath.parent.mkdir(parents=True, exist_ok=True)
filteredData.to_csv(filepath)
```
